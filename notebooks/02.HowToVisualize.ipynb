To make it easier for later development on other visualization software we use Registry Pattern. For now we implemented visualization of a full scene with rerun software. Below code registers available RerunVisualizer and opens rerun window with first scene in dataset.

```python
from darts_devkit import DARTS, VisualizeRegistry
import darts_devkit.visualization as ve

darts = DARTS("/data/darts", "test")

ve.register_rerun_visualizer()
visualizer_cls = VisualizeRegistry.get("RerunVisualizer")
visualizer = visualizer_cls()
visualizer.visualize(darts, darts.scene.all()[0].token)
```
